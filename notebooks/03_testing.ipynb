##  PDF text vs HTML text
Read the same page from the PDF to see why HTML comes first: no section tags, no LaTeX, layout noise.


```python
import httpx, pymupdf
r = httpx.get("https://arxiv.org/pdf/1706.03762", follow_redirects=True)
doc = pymupdf.open(stream=r.content, filetype="pdf")
pdf_text = doc[2].get_text()
print(len(doc), "pages")
print(pdf_text[:1500])
```
