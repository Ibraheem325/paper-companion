```python
import httpx

URL = "http://127.0.0.1:9931/v1/chat/completions"

system = "Answer only from the context. After each fact, cite its section in brackets, like [Abstract]. If the context does not contain the answer, say you do not know."

user = """Context:
[Abstract] To enable experimentation, we address two main challenges: First, we overcome a lack of high-quality datasets to benchmark table generation by curating and releasing arxivDIGESTables, a new dataset of 2,228 literature review tables extracted from ArXiv papers that synthesize a total of 7,542 research papers.

Question: What dataset does this paper introduce, and how many tables does it contain?

End every sentence with its section in brackets, for example: The model has 6 layers [Model Architecture]."""

messages = [
    {"role": "system", "content": system},
    {"role": "user", "content": user},
]

response = httpx.post(URL, json={"messages": messages}, timeout=60)
print(response.json()["choices"][0]["message"]["content"])
```

```python
from paper_companion.index.vectors import load_index
from paper_companion.retrieve.search import search

question = "How many layers does the encoder have?"

chunks, vectors = load_index("1706.03762")
hits = search(question, chunks, vectors, k=3)

context = "\n\n".join(f"[{h['section']}] {h['text']}" for h in hits)

user = f"""Context:
{context}

Question: {question}

End every sentence with its section in brackets, for example: The model has 6 layers [Model Architecture]."""

messages = [
    {"role": "system", "content": system},
    {"role": "user", "content": user},
]

response = httpx.post(URL, json={"messages": messages}, timeout=60)
for h in hits:
    print(round(h["score"], 3), h["section"])
print()
print(response.json()["choices"][0]["message"]["content"])
```
